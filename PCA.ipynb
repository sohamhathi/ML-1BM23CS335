```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("heart.csv")

print(df.head())
```

```python
X = df.drop("HeartDisease", axis=1)
y = df["HeartDisease"]
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Binary columns
X['Sex'] = le.fit_transform(X['Sex'])              # M/F → 1/0
X['ExerciseAngina'] = le.fit_transform(X['ExerciseAngina'])  # Y/N → 1/0
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

multi_cat_cols = ['ChestPainType', 'RestingECG', 'ST_Slope']

ct = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first'), multi_cat_cols)
    ],
    remainder='passthrough'
)

X = ct.fit_transform(X)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Create models
svm_model = SVC()
lr_model = LogisticRegression(max_iter=1000)
rf_model = RandomForestClassifier()

# Train models
svm_model.fit(X_train, y_train)
lr_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)

# Predictions
svm_pred = svm_model.predict(X_test)
lr_pred = lr_model.predict(X_test)
rf_pred = rf_model.predict(X_test)

# Accuracy
svm_acc = accuracy_score(y_test, svm_pred)
lr_acc = accuracy_score(y_test, lr_pred)
rf_acc = accuracy_score(y_test, rf_pred)

print("SVM Accuracy:", svm_acc)
print("Logistic Regression Accuracy:", lr_acc)
print("Random Forest Accuracy:", rf_acc)
```

```python
from sklearn.decomposition import PCA

# Keep 95% of variance
pca = PCA(n_components=0.95)

X_pca = pca.fit_transform(X_scaled)

print("Original number of features:", X_scaled.shape[1])
print("Reduced number of features:", X_pca.shape[1])
```

```python
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split

# Apply PCA
pca = PCA(n_components=0.95)
X_pca = pca.fit_transform(X_scaled)

# Split again
X_train_pca, X_test_pca, y_train, y_test = train_test_split(
    X_pca, y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Create models again (safe way)
svm_model = SVC()
lr_model = LogisticRegression(max_iter=1000)
rf_model = RandomForestClassifier()

# Train on PCA data
svm_model.fit(X_train_pca, y_train)
lr_model.fit(X_train_pca, y_train)
rf_model.fit(X_train_pca, y_train)

# Predictions
svm_pred_pca = svm_model.predict(X_test_pca)
lr_pred_pca = lr_model.predict(X_test_pca)
rf_pred_pca = rf_model.predict(X_test_pca)

# Accuracy
svm_acc_pca = accuracy_score(y_test, svm_pred_pca)
lr_acc_pca = accuracy_score(y_test, lr_pred_pca)
rf_acc_pca = accuracy_score(y_test, rf_pred_pca)

print("=== Accuracy WITH PCA ===")
print("SVM:", svm_acc_pca)
print("Logistic Regression:", lr_acc_pca)
print("Random Forest:", rf_acc_pca)
```

```python
print("\n=== FINAL COMPARISON ===")

print("SVM:")
print("Without PCA:", svm_acc)
print("With PCA   :", svm_acc_pca)

print("\nLogistic Regression:")
print("Without PCA:", lr_acc)
print("With PCA   :", lr_acc_pca)

print("\nRandom Forest:")
print("Without PCA:", rf_acc)
print("With PCA   :", rf_acc_pca)
```
